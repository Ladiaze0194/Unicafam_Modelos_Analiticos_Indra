```python
presupuesto = 120000
precio_cafe = 5000
cafes_semana = 15
#1. Cuanto Gastas al mes?
Gasto_mes = precio_cafe * cafes_semana
print ("Gasto en un mes: ", Gasto_mes)
#2. ¿Cuáto te sobra
sobra = presupuesto - Gasto_mes
print ("Me sobra: ",sobra)
3#. ¿Te alcanza?
if sobra >= 0:
  print ("Me alcanza :)")
else:
  print ("No me alcanza :(")

```
