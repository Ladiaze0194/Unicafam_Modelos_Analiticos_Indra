```python
# Definimos las reglas
def Calificaciones(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota <= 4.4:
        return "Aprueba"
    else:
        return "Excelente"

# Listado de notas
notas = [2.5, 3.8, 4.7, 5.2, 3.0]

aprobados = 0
# Se agrega la función for
for nota in notas:
    estado = Calificaciones(nota)

    print(f"Nota: {nota} -> {estado}")

    if estado in ["Aprueba", "Excelente"]:
        aprobados += 1
print ("\n=== ===")
print("Total de estudiantes que aprobaron:", aprobados)
```

*Ejercicio 2*

```python
#Generación del Diccionario
carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}

subtotal = 0

for producto, datos in carrito.items():
    precio, cantidad = datos
    subtotal += precio * cantidad
print(subtotal)
```

```python
# Descuento del 10% si supera los 200.000
subtotal_con_descuento = subtotal

if subtotal > 200000:
    subtotal_con_descuento = subtotal * 0.90
    print(subtotal_con_descuento)
```

```python
# agregamos el IVA del 19%
total = subtotal_con_descuento * 1.19

print(f"Total a pagar: ${total:,.0f}")
```

*Ejercicio 3*

```python
import pandas as pd

# DataFrame inicial con 3 empleados en nómina
empleados = pd.DataFrame({
    "nombre": ["Laura", "Camila", "Jesus"],
    "area": ["Ventas", "gerencia", "contabilidad"],
    "salario": [3000000, 4500000, 3500000]
})
print("=== TABLA DE EMPLEADOS ===")
print(empleados)
```

```python
# Se agregan 3 nuevos empleados al Dataframe
nuevos = pd.DataFrame({
    "nombre": ["Tatiana", "Noah", "Juan"],
    "area": ["Ventas", "gerencia", "contabilidad"],
    "salario": [3500000, 5000000, 3000000]
})

empleados = pd.concat([empleados, nuevos], ignore_index=True)

# IMPRESIONES
print("=== TABLA DE EMPLEADOS CON LOS NUEVOS EMPLEADOS ===")
print(empleados)

```

```python
# Crear el bono
empleados["bono"] = empleados.apply(
    lambda fila: fila["salario"] * 0.10
    if fila["area"] == "Ventas"
    else fila["salario"] * 0.05,
    axis=1
)

# Costo total de nómina
empleados["costo_total"] = empleados["salario"] + empleados["bono"]

# Nómina por área
nomina_area = empleados.groupby("area")["costo_total"].sum()




print("=== TABLA DE EMPLEADOS CON LOS NUEVOS EMPLEADOS ===")
print(empleados)
print("\n=== COSTO TOTAL DE NOMINA POR AREA ===")
print(nomina_area)


```

```python
# Empleado con mayor salario
print("\n=== EMPLEADO CON MAYOR SALARIO ===")
mayor_salario = empleados.loc[empleados["salario"].idxmax()]
print("Nombre:", mayor_salario["nombre"])
print("Área:", mayor_salario["area"])
print("Salario:", mayor_salario["salario"])
```
